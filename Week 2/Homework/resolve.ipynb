```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("employee.csv", encoding="utf-8")
df = df[10:20]

plt.rcParams["font.sans-serif"] = "SimHei"
plt.figure(figsize=(8, 5))

plt.plot(df["指标"], df["第一产业就业人员(万人)"], marker="o", color="r", label="第一产业就业人员")
plt.plot(df["指标"], df["第二产业就业人员(万人)"], marker="*", color="g", label="第二产业就业人员")
plt.plot(df["指标"], df["第三产业就业人员(万人)"], marker="s", color="b", label="第三产业就业人员")

plt.legend(loc="upper left")
plt.xlabel("年份")
plt.ylabel("人数（万人）")
plt.title("2010-2019年中国各产业人员变化折线图")
plt.show()
```

```python
import numpy as np

df = pd.read_csv("GDP.csv", encoding="gbk")
df = df[df["Country"].isin(["中国", "美国"])]
m, width = np.arange(len(df)), 1 / 6

plt.rcParams["font.sans-serif"] = "SimHei"
plt.figure(figsize=(10, 7))

for ind, key in enumerate(["2018", "2019", "2020", "2021", "2022"]):
    y = df[key] / 100_000_000
    x = m + width * ind
    plt.bar(x, y, width=width, label=key + "年")
    for a, b in zip(x, y):
        plt.text(a, b, f"{b:.0f}", ha="center", va="bottom", fontsize=14)

plt.xticks(m + 2 * width, df["Country"])
plt.yticks(range(100000, 280000, 20000))
plt.ylim(100000, 270000)
plt.ylabel("美元（亿）")
plt.title("2018-2022年中美两国GDP对比柱状图")
plt.legend(loc="upper left")
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt



df = pd.read_csv("学生运动数据.csv")
df["datetime"] = pd.to_datetime(df["datetime"])
df = df[df["date"] == "2023-06-03"]
df = df.sort_values("datetime")



plt.rcParams["font.sans-serif"] = "SimHei"
fig = plt.figure(figsize=(13, 5))
ax1 = fig.add_subplot()
ax2 = ax1.twinx()
ax3 = ax1.twinx()
ax4 = ax1.twinx()

p1, = ax1.plot(df["datetime"], df["heartRate"], marker="o", color="r", label="心率")
p2, = ax2.plot(df["datetime"], df["step"], marker="o", color="g", label="步数")
p3, = ax3.plot(df["datetime"], df["distance"], marker="o", color="b", label="距离")
p4, = ax4.plot(df["datetime"], df["calories"], marker="o", color="c", label="卡路里")

ax1.set(ylim=(40, 200), xlabel="日期", ylabel="心率（次）")
ax2.set(ylim=(0, 2000), ylabel="步数（步）")
ax3.set(ylim=(0, 1.6), ylabel="距离（公里）")
ax4.set(ylim=(0, 100), ylabel="卡路里（千卡）")

ax3.spines.right.set_position(("axes", 1.1))
ax4.spines.right.set_position(("axes", 1.2))
ax1.yaxis.label.set_color(p1.get_color())
ax2.yaxis.label.set_color(p2.get_color())
ax3.yaxis.label.set_color(p3.get_color())
ax4.yaxis.label.set_color(p4.get_color())
ax4.spines.left.set_color(p1.get_color())
ax2.spines.right.set_color(p2.get_color())
ax3.spines.right.set_color(p3.get_color())
ax4.spines.right.set_color(p4.get_color())
ax1.tick_params(axis="y", colors=p1.get_color())
ax2.tick_params(axis="y", colors=p2.get_color())
ax3.tick_params(axis="y", colors=p3.get_color())
ax4.tick_params(axis="y", colors=p4.get_color())

plt.legend(handles=[p1, p2, p3, p4], loc="upper right")
plt.title("学生运动相关数据变化折线图")
plt.tight_layout()
plt.show()
```
